```python
# Exercise 3
# Indexing & Slicing

# FYDP Table 5.21
temperatures = [320, 333, 345, 354, 365]
evaporation_rates = [0.45, 0.65, 0.82, 1.00, 1.28]
freshwater_yields = [0.40, 0.58, 0.75, 1.00, 1.20]

# Indexing
print("Indexing")
print(f"First temperature: {temperatures[0]} K")
print(f"Last temperature: {temperatures[-1]} K")
print(f"Base temperature: {temperatures[3]} K")

# Slicing
print(f"\nSlicing ")
print(f"First 3 temperatures: {temperatures[0:3]}")
print(f"Last 2 temperatures: {temperatures[-2:]}")
print(f"Middle values: {temperatures[1:4]}")
```

```python
# Exercise 4
# Complete FYDP Example: List Methods

temperatures = [320, 333, 345, 354, 365]
evaporation_rates = [0.45, 0.65, 0.82, 1.00, 1.28]

print("Original Data")
print(f"Temperatures: {temperatures}")
print(f"Evaporation Rates: {evaporation_rates}")

# highest evaporation rate index
max_evap = max(evaporation_rates)
max_index = evaporation_rates.index(max_evap)
print(f"\nMax Evaporation")
print(f"Max rate: {max_evap} at index {max_index}")
print(f"Corresponding temperature: {temperatures[max_index]} K")

# new observation add
print(f"\nAdding New Observation")
temperatures.insert(0, 310)
evaporation_rates.insert(0, 0.35)
print(f"After adding 310K: {temperatures}")
print(f"After adding 0.35: {evaporation_rates}")

# one observation remove
print(f"\nRemoving Observation")
temperatures.remove(310)
evaporation_rates.remove(0.35)
print(f"After removing 310K: {temperatures}")
print(f"After removing 0.35: {evaporation_rates}")

# last observation remove
print(f"\nLast Observation")
last_temp = temperatures.pop()
last_evap = evaporation_rates.pop()
print(f"Removed: {last_temp}K, {last_evap}")
print(f"Final temperatures: {temperatures}")
print(f"Final evaporation rates: {evaporation_rates}")

# how many times a value is in the list
print(f"\nCounting")
print(f"345 appears {temperatures.count(345)} time(s)")
```

```python
# Exercise 5
# Tuple Examples from FYDP Dataset

# observation tuple
obs1 = (320, 0.45, 0.40)
obs2 = (333, 0.65, 0.58)
obs3 = (345, 0.82, 0.75)
obs4 = (354, 1.00, 1.00)
obs5 = (365, 1.28, 1.20)

print("Individual Observations (Tuples)")
print(f"Observation 1: {obs1}")
print(f"Observation 2: {obs2}")

# Tuple indexing
print(f"Obs1 Temperature: {obs1[0]} K")
print(f"Obs1 Evaporation: {obs1[1]}")
print(f"Obs1 Yield: {obs1[2]}")

# Tuple unpacking
temp, evap, yield_ = obs3
print(f"Unpacked Obs3: Temp={temp}K, Evap={evap}, Yield={yield_}")

# List of Tuples
dataset = [obs1, obs2, obs3, obs4, obs5]

print("\nFull Dataset")
for i, (temp, evap, yield_) in enumerate(dataset):
    print(f"Obs {i+1}: Temp={temp}K, Evap={evap}, Yield={yield_}")

# .count() এবং .index() in touple
print(f"\nCount of 1.00 in obs4: {obs4.count(1.00)}")
print(f"Index of 0.82 in obs3: {obs3.index(0.82)}")
```
