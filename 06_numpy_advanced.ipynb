BROADCASTING

```python
import numpy as np

# Scalar + Array
a = np.array([1, 2, 3])
print(a + 10)

# 1D + 2D
matrix = np.array([[1, 2, 3], [4, 5, 6]])
row = np.array([10, 20, 30])
print(matrix + row)

# Column + Row
col = np.array([[1], [2]])
print(matrix + col)
```

Linear Algebra ( Need for Linear Regression)

```python
A = np.array([[1, 2], [3, 4]])

# Determinant
print(np.linalg.det(A))

# Inverse
print(np.linalg.inv(A))

# Eigenvalues
eigenvalues, eigenvectors = np.linalg.eig(A)
print(eigenvalues)

# Matrix multiplication (Dot Product)
B = np.array([[5, 6], [7, 8]])
print(np.dot(A, B))
```

Normal Equation (Linear Regression)

```python
# θ = (X^T X)^-1 X^T y
X = np.array([[1, 320], [1, 333], [1, 345], [1, 354], [1, 365]])
y = np.array([0.40, 0.58, 0.75, 1.00, 1.20])

theta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
print(f"Theta: {theta}")
# [intercept, slope]
```

```python
# FYDP DATASETS
# NumPy Advanced Practice

import numpy as np

# Broadcasting
temperatures = np.array([320, 333, 345, 354, 365])
normalized = (temperatures - np.mean(temperatures)) / np.std(temperatures)
print(f"Normalized: {normalized}")

# Random for ML
np.random.seed(42)
shuffled = np.random.permutation(temperatures)
print(f"Shuffled: {shuffled}")

# Linear Algebra
X = np.array([[1, 1], [1, 2], [1, 3]])
y = np.array([1, 2, 3])
theta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
print(f"Theta: {theta}")

# Boolean Masking
high_temp = temperatures[temperatures > 340]
print(f"High temps: {high_temp}")

# np.where()
labels = np.where(temperatures > 340, "High", "Low")
print(f"Labels: {labels}")
```
