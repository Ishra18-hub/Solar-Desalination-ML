```python
# Creating CSV FILE (desalination_data.csv) from FYDP Report

# Import libraries
import pandas as pd
from google.colab import files
```

```python
# 1. Temperature vs Evaporation Rate vs Freshwater Yield (Table 5.21)

temp_data = {
    'Temperature_K': [320, 333, 345, 354, 365],
    'Evaporation_Rate_relative': [0.45, 0.65, 0.82, 1.00, 1.28],
    'Freshwater_Yield_relative': [0.40, 0.58, 0.75, 1.00, 1.20]
}

df_temp = pd.DataFrame(temp_data)
print(f"\nTemperature Data")
print(df_temp)
```

```python
# 2. Air Flow Rate vs Performance (Table 5.22)

airflow_data = {
    'Air_Flow_ms': [5, 8, 10, 12, 15],
    'Evaporation_Rate_relative': [0.72, 0.92, 1.00, 0.98, 0.95],
    'Condensation_Efficiency': [0.82, 0.88, 0.87, 0.80, 0.72]
}

df_airflow = pd.DataFrame(airflow_data)
print("Air Flow Data")
print(df_airflow)
```

```python
# 3. Droplet Size vs Evaporation (Table 5.23)

droplet_data = {
    'Droplet_Size_um': [500, 300, 200, 100, 50],
    'Surface_Area_Ratio_relative': [0.12, 0.20, 0.30, 0.60, 1.00],
    'Evaporation_Rate_relative': [0.38, 0.55, 0.72, 0.89, 1.00]
}

df_droplet = pd.DataFrame(droplet_data)
print("Droplet Size Data")
print(df_droplet)
```

```python
# 4. Solar Irradiance vs Yield (Table 5.24)

solar_data = {
    'Solar_Irradiance_Wm2': [400, 600, 800, 1000, 1200],
    'Achievable_Inlet_Temp_K': [330, 340, 348, 354, 360],
    'Freshwater_Yield_relative': [0.48, 0.62, 0.78, 1.00, 1.18]
}

df_solar = pd.DataFrame(solar_data)
print("Solar Irradiance Data")
print(df_solar)
```

```python
# 5. Ambient Temperature vs Performance (Table 5.25)

ambient_data = {
    'Ambient_Temperature_K': [285, 290, 295, 300, 305],
    'Condensation_Efficiency': [0.92, 0.90, 0.88, 0.87, 0.83],
    'Heat_Loss': [0.18, 0.20, 0.22, 0.24, 0.27]
}

df_ambient = pd.DataFrame(ambient_data)
print("Ambient Temperature Data")
print(df_ambient)
```

```python
# 6. Mass Balance Data (Fig 5.11 & 5.12 / from the simulation results)

mass_balance = {
    'Parameter': ['Inlet', 'Freshwater', 'Brine', 'Losses'],
    'Value_kg_hr': [20.4, 3.672, 15.328, 1.4],
    'Percentage': [100.0, 18.0, 75.14, 6.86]
}

df_mass = pd.DataFrame(mass_balance)
print("Mass Balance Data")
print(df_mass)
```

```python
# 7. Economic Data (Table 5.27)
economic_data = {
    'Parameter': [
        'Capital Cost (BDT)', 'Freshwater Production (L/day)',
        'O&M Cost (BDT/year)', 'System Lifetime (years)',
        'LCOW (BDT/L)', 'LCOW (USD/m3)'
    ],
    'Min_Value': [35000, 10, 2000, 10, 0.8, 8],
    'Max_Value': [50000, 15, 4000, 15, 1.2, 12]
}

df_economic = pd.DataFrame(economic_data)
print("Economic Data")
print(df_economic)
```

```python
# Create FULL Dataset for ML

# Important dataset for ML
# Temperature + Air Flow + Solar Irradiance + Ambient Temp → Freshwater Yield

# Base: Temperature data (5 rows)
# we will add different feature's typical values

main_data = {
    'Temperature_K': [320, 333, 345, 354, 365],
    'Air_Flow_ms': [5, 8, 10, 12, 15],
    'Solar_Irradiance_Wm2': [400, 600, 800, 1000, 1200],
    'Ambient_Temperature_K': [285, 290, 295, 300, 305],
    'Droplet_Size_um': [500, 300, 200, 100, 50],
    'Evaporation_Rate_relative': [0.45, 0.65, 0.82, 1.00, 1.28],
    'Freshwater_Yield_relative': [0.40, 0.58, 0.75, 1.00, 1.20],
    'Condensation_Efficiency': [0.92, 0.90, 0.88, 0.87, 0.83],
    'Heat_Loss': [0.18, 0.20, 0.22, 0.24, 0.27]
}

df_main = pd.DataFrame(main_data)

print("Main Dataset")
print(df_main)
print(f"\nShape: {df_main.shape}")
print(f"Columns: {list(df_main.columns)}")
```

```python
# Save to CSV

# Save main dataset
df_main.to_csv('desalination_data.csv', index=False)
print("desalination_data.csv created successfully")

# Save individual tables (for reference)
df_temp.to_csv('table_5_21_temperature.csv', index=False)
df_airflow.to_csv('table_5_22_airflow.csv', index=False)
df_droplet.to_csv('table_5_23_droplet.csv', index=False)
df_solar.to_csv('table_5_24_solar.csv', index=False)
df_ambient.to_csv('table_5_25_ambient.csv', index=False)
df_mass.to_csv('table_5_11_mass_balance.csv', index=False)
df_economic.to_csv('table_5_27_economic.csv', index=False)

print("All individual tables are saved")
```
