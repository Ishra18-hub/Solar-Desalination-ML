```python
# Exercise 3
# Using sign() Function ( kaggle exercise)

def sign(x):
    """Return -1, 0, or 1 based on the sign of x."""
    if x < 0:
        return -1
    elif x > 0:
        return 1
    else:
        return 0

print(sign(-5))
print(sign(0))
print(sign(10))
```

```python
# Exercise 4: Boolean Conversion

# with int() Boolean transform in 0/1
conditions = [True, False, True, True, False]
count = sum(int(c) for c in conditions)
print(f"True count: {count}")

# FYDP
temperatures = [320, 333, 345, 354, 365]
good_temp = [temp > 340 for temp in temperatures]
print(f"\nGood temperature observations: {good_temp}")
print(f"Count: {sum(int(c) for c in good_temp)}/5")
```

```python
# Exercise 5: FYDP Practical Application

data = [
    (320, 500, 5, 0.40),
    (333, 600, 7, 0.58),
    (345, 850, 9, 0.75),
    (354, 1000, 10, 1.00),
    (365, 1200, 12, 1.20)
]

print("Optimal Conditions Check")
for i, (temp, irr, air, yield_) in enumerate(data):
    good_temp = temp >= 345
    good_irr = irr >= 800
    good_air = air >= 8

    count = int(good_temp) + int(good_irr) + int(good_air)

    if count == 3:
        status = "ALL conditions met"
    elif count == 2:
        status = "2/3 conditions met"
    else:
        status = "Not optimal"

    print(f"Obs {i+1}: Temp={temp}K, Yield={yield_} = {status}")
```

```python
# Exercise 6: Exactly One Condition (XOR)

def exactly_one_optimal(temp, irradiance, air_flow):
    """
    Return True if exactly ONE condition is optimal.
    Optimal: temp >= 354, irradiance >= 1000, air_flow >= 10
    """
    good_temp = temp >= 354
    good_irr = irradiance >= 1000
    good_air = air_flow >= 10

    return (int(good_temp) + int(good_irr) + int(good_air)) == 1

# Test
print("Exactly One Optimal")
print(f"354, 500, 5 = {exactly_one_optimal(354, 500, 5)}")
print(f"354, 1000, 5 = {exactly_one_optimal(354, 1000, 5)}")
print(f"320, 500, 10 = {exactly_one_optimal(320, 500, 10)}")
print(f"354, 1000, 10 = {exactly_one_optimal(354, 1000, 10)}")
```
