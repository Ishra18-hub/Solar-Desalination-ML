```python
# Author: Ishra Ismat Kamal
# Date: 19 September 2026
# Topic: Pandas Filtering with FYDP Data

import pandas as pd
import numpy as np

# Load FYDP data
from google.colab import files
uploaded = files.upload()

df = pd.read_csv('desalination_data.csv')
print("FYDP Data")
print(df)
```

 Boolean Indexing

```python
# Boolean Indexing - High Temperature
print("High Temperature (>350 K)")
high_temp = df[df['Temperature_K'] > 350]
print(high_temp)
```

Multiple Conditions (AND)

```python
# AND Condition
print("High Temp AND High Yield")
result = df[
    (df['Temperature_K'] > 350) &
    (df['Freshwater_Yield_relative'] > 0.9)
]
print(result)
```

Multiple Conditions (OR)

```python
# OR Condition
print("High Temp OR High Yield")
result = df[
    (df['Temperature_K'] > 350) |
    (df['Freshwater_Yield_relative'] > 1.0)
]
print(result)
```

.loc- Label-based

```python
# .loc - specific columns
print("Selected Columns (loc)")
result = df.loc[df['Temperature_K'] > 350,
                ['Temperature_K', 'Freshwater_Yield_relative']]
print(result)
```

.iloc- Position-based

```python
# .iloc - first 3 rows, first 3 columns
print("First 3 Rows (iloc)")
result = df.iloc[0:3, 0:3]
print(result)
```

.query()- String-based

```python
# .query() - string condition
print("Query: High Temp and High Yield")
result = df.query('Temperature_K > 350 and Freshwater_Yield_relative > 0.9')
print(result)
```

.isin()- List-based

```python
# .isin() - specific values
print("Specific Temperatures (320, 365)")
result = df[df['Temperature_K'].isin([320, 365])]
print(result)
```

.between()- Range-based

```python
# .between() - range filter
print("Temperature Between 330 and 360")
result = df[df['Temperature_K'].between(330, 360)]
print(result)
```

Practice — Combined filtering

```python
# Combined filtering practice
print("Combined: Low Temp AND High Heat Loss")
result = df[
    (df['Temperature_K'] < 340) &
    (df['Heat_Loss'] > 0.2)
]
print(result)
```

## Note on Empty Result
This filter (Temp < 340 AND Heat_Loss > 0.2) returned no rows because
in the FYDP data, low temperature conditions are associated with
lower heat loss values. This demonstrates that filtering with strict
conditions can return empty results which is a valid outcome.

Saving filtered results

```python
# Save filtered results
high_temp_data = df[df['Temperature_K'] > 350]
high_temp_data.to_csv('high_temp_data.csv', index=False)
print("Saved: high_temp_data.csv")
```
