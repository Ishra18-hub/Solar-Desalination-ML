Slicing

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5])
print(arr[4:])
print(arr[:4])
print(arr[-3:-1])  # negative slicing
print(arr[1:5:2])  # using the step value [start:end:step]
print(arr[::2])
```

Statistical Operations

```python
temperatures = np.array([320, 333, 345, 354, 365])

print(f"Mean: {np.mean(temperatures)}")
print(f"Median: {np.median(temperatures)}")
print(f"Std: {np.std(temperatures)}")         # Standard Deviation
print(f"Max: {np.max(temperatures)}")
print(f"Min: {np.min(temperatures)}")
print(f"Sum: {np.sum(temperatures)}")
```

Dot Product (For Linear Regression)

```python
# FYDP Efficiency Calculation
# η = m * hfg / (I * A * t)

m = np.array([0.001, 0.002, 0.003])      # mass flow rate
hfg = 2260000                              # latent heat
I = np.array([1000, 1000, 1000])          # irradiance
A = 1                                      # area
t = 3600                                   # time

efficiency = (m * hfg) / (I * A * t)
print(f"Efficiency: {efficiency}")
```

Matrix Operations

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(np.dot(A, B))       # Matrix multiplication
print(A.T)                # Transpose
print(np.linalg.inv(A))   # Inverse
```

Exercise 3

```python
# FYDP data

data = np.array([
    [320, 0.45, 0.40, 5],
    [333, 0.65, 0.58, 8],
    [345, 0.82, 0.75, 10],
    [354, 1.00, 1.00, 12],
    [365, 1.28, 1.20, 15]
])
# Columns: Temperature, Evaporation, Yield, Air_Flow

# Extract Temperature column
temp = data[:, 0]
print(f"Temperatures: {temp}")

# Calculate mean temperature
print(f"Mean Temperature: {np.mean(temp):.2f} K")

# Extract Yield column
yield_ = data[:, 2]
print(f"Yields: {yield_}")

# Find max yield
max_yield = np.max(yield_)
max_index = np.argmax(yield_)
print(f"Max Yield: {max_yield} at Temperature {temp[max_index]} K")

# Filter high temperature observations
high_temp = data[data[:, 0] > 340]
print(f"High temperature data:\n{high_temp}")
```
