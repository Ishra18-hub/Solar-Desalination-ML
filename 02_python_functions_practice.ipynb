```python
# Exercise 1: def and return
# docstring (right after def fuction called)

def calculate_efficiency(m, hfg, I, A, t):
    """
    Calculate thermal efficiency of solar desalination system.
    Formula: η = (m * hfg) / (I * A * t)
    """
    efficiency = (m * hfg) / (I * A * t)
    return efficiency

# Test with FYDP values
m = 0.001        # mass flow rate (kg/s)
hfg = 2260000    # latent heat of vaporization (J/kg)
I = 1000         # solar irradiance (W/m²)
A = 1            # collector area (m²)
t = 3600         # time (s)

eff = calculate_efficiency(m, hfg, I, A, t)
print(f"Thermal Efficiency: {eff:.4f} ({eff*100:.2f}%)")
```

```python
# Exercise 3: Default Arguments

def predict_yield(temperature, irradiance, efficiency=0.32):
    """
    Predict freshwater yield based on temperature and irradiance.
    Efficiency default = 0.32 (32.35% from FYDP)
    """
    yield_kg_hr = (temperature / 354) * (irradiance / 1000) * efficiency * 20.4
    return yield_kg_hr

# Default efficiency (0.32)
yield1 = predict_yield(354, 1000)
print(f"Yield with default efficiency: {yield1:.2f} kg/hr")

# with Custom efficiency
yield2 = predict_yield(354, 1000, efficiency=0.25)
print(f"Yield with 25% efficiency: {yield2:.2f} kg/hr")
```

```python
# Exercise 5: key argument (max, min, sorted)

# FYDP data
data = [
    {"temp": 320, "evap": 0.45, "yield": 0.40},
    {"temp": 333, "evap": 0.65, "yield": 0.58},
    {"temp": 345, "evap": 0.82, "yield": 0.75},
    {"temp": 354, "evap": 1.00, "yield": 1.00},
    {"temp": 365, "evap": 1.28, "yield": 1.20},
]

# max() - key argument
best = max(data, key=lambda x: x["yield"])
print(f"Best yield: {best}")

# sorted() - key argument
sorted_data = sorted(data, key=lambda x: x["temp"])
print(f"Sorted by temperature:")
for d in sorted_data:
    print(f"  {d['temp']}K → Yield: {d['yield']}")
```

```python
# Exercise 6: Higher-order Functions

def apply_to_all(data_list, func):
    """Apply a function to all items in a list."""
    return [func(item) for item in data_list]

# Custom function
def square(x):
    return x ** 2

# Using Higher-order function
numbers = [1, 2, 3, 4, 5]
squared = apply_to_all(numbers, square)
print(f"Original: {numbers}")
print(f"Squared: {squared}")

# Lambda (anonymous function)
cubed = apply_to_all(numbers, lambda x: x ** 3)
print(f"Cubed: {cubed}")
```
