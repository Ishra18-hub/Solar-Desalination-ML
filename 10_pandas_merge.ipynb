```python
# Author: Ishra Ismat Kamal
# Date: 19 September 2026
# Topic: Pandas Merge and Apply with FYDP Data

import pandas as pd
import numpy as np

# Load FYDP data
from google.colab import files
uploaded = files.upload()

df = pd.read_csv('desalination_data.csv')
print("FYDP Data")
print(df)
```

SPLIT AND MERGE

```python
# Split data into inputs and outputs
inputs = df[['Temperature_K', 'Air_Flow_ms', 'Solar_Irradiance_Wm2',
             'Ambient_Temperature_K', 'Droplet_Size_um']].copy()
inputs['Sample_ID'] = range(1, 6)

outputs = df[['Evaporation_Rate_relative', 'Freshwater_Yield_relative',
              'Condensation_Efficiency', 'Heat_Loss']].copy()
outputs['Sample_ID'] = range(1, 6)

# Merge
merged = pd.merge(inputs, outputs, on='Sample_ID')
print("\nMerged Data")
print(merged)
```

apply() function

```python
def categorize_temp(temp):
    if temp < 335:
        return 'Low'
    elif temp < 355:
        return 'Medium'
    else:
        return 'High'

df['Temp_Category'] = df['Temperature_K'].apply(categorize_temp)
print("\nTemperature Categories")
print(df[['Temperature_K', 'Temp_Category']])
```

Lambda with apply

```python
# Efficiency score
df['Efficiency_Score'] = df.apply(
    lambda row: row['Freshwater_Yield_relative'] / row['Evaporation_Rate_relative'],
    axis=1
)
print("\nEfficiency Scores")
print(df[['Temperature_K', 'Efficiency_Score']])
```

Efficiency_Score is a derived metric for demonstration only.
It is **NOT** the same as the FYDP thermal efficiency (32.35%).
This is calculated as Yield/Evaporation Rate for practice purposes.
