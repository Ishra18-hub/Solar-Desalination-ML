```python
# Exercise 1: Variables & Data Types

temperature = 354          # int (TEMP. Kelvin)
evaporation_rate = 0.82    # float
system_name = "Solar Desalination"  # string
is_efficient = True        # bool

print("=== FYDP Variables ===")  #structure : f"string:{variable}
print(f"Temperature: {temperature} K")
print(f"Evaporation Rate: {evaporation_rate}")
print(f"System: {system_name}")
print(f"Is Efficient? {is_efficient}")

# Arithmetic Practice
inlet_temp = 354
outlet_temp = 342
temp_difference = inlet_temp - outlet_temp
print(f"\nTemperature Difference: {temp_difference} K") #\n means new line
```

```python
# Exercise 2: Lists & Loops

# collected from FYDP's Table 5.21
temperatures = [320, 333, 345, 354, 365]

# temp. and evaporation rate list using for loop
print("==Temperature List==")
for temp in temperatures:
    print(f"Temperature: {temp} K")

evaporation_rates = [0.45, 0.65, 0.82, 1.00, 1.28]

print("\n=== Temperature vs Evaporation Rate ===")
for temp, evap in zip(temperatures, evaporation_rates):
    print(f"{temp} K → {evap} (relative)")
```

```python
# Exercise 3: Dictionary

# for observation
observation = {
    "temperature": 354,
    "evaporation_rate": 1.00,
    "freshwater_yield": 1.00,
    "air_flow": 10,
    "solar_irradiance": 1000
}

print("== Observation Data ==")
for key, value in observation.items():
    print(f"{key}: {value}")

# a specific value
print(f"\nFreshwater Yield: {observation['freshwater_yield']}")
```
