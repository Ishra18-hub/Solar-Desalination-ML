Setup + Data Load

```python
# Author: Ishra Ismat Kamal
# Date: 25 September 2026
# Topic: Matplotlib Basics with FYDP Data

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load FYDP data
from google.colab import files
uploaded = files.upload()

df = pd.read_csv('desalination_data.csv')
print("FYDP Data")
print(df)
print(f"\nShape: {df.shape}")
```

Line plot

```python
# Line Plot
plt.figure(figsize=(8, 5))
plt.plot(df['Temperature_K'], df['Freshwater_Yield_relative'],
         marker='o', color='orange', linewidth=2, markersize=10)
plt.xlabel('Temperature (K)', fontsize=12)
plt.ylabel('Freshwater Yield (relative)', fontsize=12)
plt.title('Temperature vs Freshwater Yield', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('line_plot_temp_vs_yield.png', dpi=300)
plt.show()
print("Saved: line_plot_temp_vs_yield.png")
```

Scatter Plot

```python
# Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(df['Solar_Irradiance_Wm2'], df['Freshwater_Yield_relative'],
            color='green', s=150, alpha=0.7, edgecolors='black')
plt.xlabel('Solar Irradiance (W/m²)', fontsize=12)
plt.ylabel('Freshwater Yield (relative)', fontsize=12)
plt.title('Solar Irradiance vs Freshwater Yield', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('scatter_plot_solar_vs_yield.png', dpi=300)
plt.show()
print("Saved: scatter_plot_solar_vs_yield.png")
```

Bar chart

```python
# Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(df['Droplet_Size_um'].astype(str), df['Freshwater_Yield_relative'],
        color='steelblue', edgecolor='black')
plt.xlabel('Droplet Size (μm)', fontsize=12)
plt.ylabel('Freshwater Yield (relative)', fontsize=12)
plt.title('Droplet Size vs Freshwater Yield', fontsize=14)
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig('bar_chart_droplet_vs_yield.png', dpi=300)
plt.show()
print("Saved: bar_chart_droplet_vs_yield.png")
```

Histogram

```python
# Histogram
plt.figure(figsize=(8, 5))
plt.hist(df['Freshwater_Yield_relative'], bins=5, color='purple',
         edgecolor='black', alpha=0.7)
plt.xlabel('Freshwater Yield (relative)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Freshwater Yield', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('histogram_yield_distribution.png', dpi=300)
plt.show()
print("Saved: histogram_yield_distribution.png")
```

Subplots

```python
# Subplots- 4 plots in one figure
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Temperature vs Yield
axes[0, 0].plot(df['Temperature_K'], df['Freshwater_Yield_relative'],
                marker='o', color='orange', linewidth=2)
axes[0, 0].set_xlabel('Temperature (K)')
axes[0, 0].set_ylabel('Freshwater Yield')
axes[0, 0].set_title('Temperature vs Yield')
axes[0, 0].grid(True, alpha=0.3)

# 2. Air Flow vs Yield
axes[0, 1].plot(df['Air_Flow_ms'], df['Freshwater_Yield_relative'],
                marker='s', color='blue', linewidth=2)
axes[0, 1].set_xlabel('Air Flow (m/s)')
axes[0, 1].set_ylabel('Freshwater Yield')
axes[0, 1].set_title('Air Flow vs Yield')
axes[0, 1].grid(True, alpha=0.3)

# 3. Solar Irradiance vs Yield
axes[1, 0].plot(df['Solar_Irradiance_Wm2'], df['Freshwater_Yield_relative'],
                marker='^', color='green', linewidth=2)
axes[1, 0].set_xlabel('Solar Irradiance (W/m²)')
axes[1, 0].set_ylabel('Freshwater Yield')
axes[1, 0].set_title('Solar Irradiance vs Yield')
axes[1, 0].grid(True, alpha=0.3)

# 4. Ambient Temperature vs Yield
axes[1, 1].plot(df['Ambient_Temperature_K'], df['Freshwater_Yield_relative'],
                marker='D', color='red', linewidth=2)
axes[1, 1].set_xlabel('Ambient Temperature (K)')
axes[1, 1].set_ylabel('Freshwater Yield')
axes[1, 1].set_title('Ambient Temperature vs Yield')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('subplots_4in1.png', dpi=300)
plt.show()
print("Saved: subplots_4in1.png")
```

Markdown Note — Visualization Limitations

## NOTE: Visualization Limitations

These plots show near-perfect linear relationships because:
1. The data was generated sequentially (320K → 365K)
2. All features change together in the CFD simulation
3. With only 5 points, any pattern looks "perfect"

**In real experimental data, we would expect:**
- Scatter around the trend line
- Noise in measurements
- Non-linear relationships

These plots demonstrate the visualization code, not scientific findings.
