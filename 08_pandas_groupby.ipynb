ARTIFICIAL DATA

```python
# Artificial data for proper GroupBy practice

np.random.seed(42)
n = 30

practice_df = pd.DataFrame({
    'Temperature_K': np.random.uniform(320, 365, n),
    'Air_Flow_ms': np.random.uniform(5, 15, n),
    'Solar_Irradiance_Wm2': np.random.uniform(400, 1200, n),
    'Ambient_Temperature_K': np.random.uniform(285, 305, n),
    'Droplet_Size_um': np.random.uniform(50, 500, n)
})

# Target: yield depends on temperature + noise
practice_df['Freshwater_Yield_relative'] = (
    0.003 * practice_df['Temperature_K'] - 0.5 +
    np.random.normal(0, 0.05, n)
)

print(" Artificial Data (30 rows) ")
print(practice_df.head())
print(f"\nTotal rows: {len(practice_df)}")
```

GROUPBY PRACTICE 1( WITH 30 ROWS)

```python
# Create temperature bins
practice_df['Temp_Range'] = pd.cut(
    practice_df['Temperature_K'],
    bins=[300, 340, 360, 400],
    labels=['Low', 'Medium', 'High']
)

# GroupBy with statistics
print(" GroupBy Temp_Range ")
result = practice_df.groupby('Temp_Range', observed=True)['Freshwater_Yield_relative'].agg(
    ['mean', 'std', 'min', 'max', 'count']
)
print(result)
```

GROUPBY PRACTICE 2

```python
# GroupBy multiple columns
practice_df['Air_Flow_Category'] = pd.cut(
    practice_df['Air_Flow_ms'],
    bins=[0, 8, 12, 20],
    labels=['Low', 'Medium', 'High']
)

print(" GroupBy Temp_Range + Air_Flow_Category ")
print(practice_df.groupby(['Temp_Range', 'Air_Flow_Category'], observed=True)['Freshwater_Yield_relative'].mean())
```
